```python
from syncode import Syncode
```

```python
from lark import Lark
grammar = r"""
start: number "," number "," operator
number: SIGNED_NUMBER
operator: "+" | "-" | "*" | "/"
%import common.SIGNED_NUMBER
%ignore " "
"""
parser = Lark(grammar)
```

```python
print(parser.parse("45.1,23.54,+").pretty())
print(parser.parse("-15.5,60.0,+").pretty())
print(parser.parse("26.0,-14.5,-").pretty())
```

```python
syn_llm = Syncode(model="microsoft/phi-2", grammar=grammar, mode="grammar_strict", parse_output_only=True, max_new_tokens=40)
```

```python
examples = """
Extract the two numbers and the operator.
Output format: first_number,second_number,operator
Do not calculate the answer.

Question: What is 18.5 plus 7.2?
Output: 18.5,7.2,+

Question: What is 25.0 minus 9.5?
Output: 25.0,9.5,-

Question: Multiply -4.0 by 3.5.
Output: -4.0,3.5,*

Question: What is 42.0 divided by 6.0?
Output: 42.0,6.0,/

Question:
"""
```

```python
question = "What is 45.1 plus 23.54?"

prompt = examples + question + "\nOutput:"
output = syn_llm.infer(prompt)[0]

print("Model output:", repr(output))
```

```python
def run_syncode(textual_query):
    prompt = examples + textual_query + "\nOutput:"
    output = syn_llm.infer(prompt)[0].strip()

    parts = output.split(",")
    a = float(parts[0])
    b = float(parts[1])
    op = parts[2].strip()

    if op == "+":
        answer = a + b
    elif op == "-":
        answer = a - b
    elif op == "*":
        answer = a * b
    elif op == "/":
        answer = a / b
    else:
        raise ValueError("Unknown operator")

    return ([a, b], op, float(answer))
```

```python
print(run_syncode("What is 327. multiplied by 11.0?"))
print(run_syncode("What is 45.1 plus 23.54?"))
print(run_syncode("What is 120.4 divided by 4.0?"))
print(run_syncode("What is negative 15.5 plus 60.0?"))
print(run_syncode("What is 14.5 subtracted from 26.0?"))
```
